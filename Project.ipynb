```python
#drop_duplicates
df.drop_duplicates(inplace =True)
```

```python
# fix columns name
df.columns =df.columns.str.lower().str.replace(' ','_')
```

```python
#fix date columns
df['order_date'] =pd.to_datetime(df['order_date'])
```

```python
# fill or drop null
df.dropna(subset=['order_id' ], inplace=True)
df['discount_percent'].fillna(0, inplace=True)
```

```python
# Run this to see exact column names
print(df.columns.tolist())
```

```python
# Fix column names → lowercase, underscores
df.columns = df.columns.str.lower().str.replace(' ', '_')

# Fix date column
df['order_date'] = pd.to_datetime(df['order_date'])

# Add calculated columns (your dataset has cost_price & list_price)
df['discount_amount'] = (df['list_price'] * df['discount_percent'] / 100).round(2)
df['sale_price']      = (df['list_price'] - df['discount_amount']).round(2)
df['profit']          = ((df['sale_price'] - df['cost_price']) * df['quantity']).round(2)
df['revenue']         = (df['sale_price'] * df['quantity']).round(2)
df['profit_margin']   = (df['profit'] / df['revenue'] * 100).round(2)

# Extract time parts
df['year']  = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.strftime('%b')

print("✅ Cleaned! Shape:", df.shape)
print(df[['order_id','sale_price','profit','revenue','profit_margin']].head())
```
